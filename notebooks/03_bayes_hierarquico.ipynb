## 1. Resumo dos resultados salvos

````python
import os
import pandas as pd
import arviz as az

RESULTS = r'C:\Users\tito\OneDrive\Documentos\Projetos\spotify_challenge\insights-spotfy-grupo-4\relatorio\analises\resultados'

print('=== Artefatos Q8 ===')
for f in sorted(os.listdir(RESULTS)):
    if f.startswith('q8_'):
        path = os.path.join(RESULTS, f)
        size = os.path.getsize(path)
        print(f'  {f} ({size/1024:.1f} KB)')
````

## 2. Efeitos globais (Gaussian e Bernoulli)

## 3. Diagnósticos

````python
for fam in ['gaussian', 'bernoulli']:
    nc = os.path.join(RESULTS, f'q8_model_{fam}.nc')
    if not os.path.exists(nc):
        print(f'{fam}: posterior nao encontrado')
        continue
    idata = az.from_netcdf(nc)
    print(f'\n=== {fam} ===')
    print(az.summary(idata, var_names=['mu_alpha', 'sigma_alpha', 'mu_beta', 'sigma_beta']))
````
